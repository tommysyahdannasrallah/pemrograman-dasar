```python
# Deklarasi tipe primitif dengan konteks kelistrikan
frekuensi = 50                 # int  — Hz
faktor_daya = 0.85             # float — cos φ
tiga_fasa = True               # bool
label = "Trafo Distribusi"     # str
impedansi = complex(4, 3)      # complex — Z = R + jX (Ohm)


print(frekuensi,type(frekuensi))
print(faktor_daya,type(faktor_daya))
print(tiga_fasa,type(tiga_fasa))
print(label,type(label))
print(impedansi,type(impedansi))
```

```python
tegangan_harian = [21.5,219.0,220.3,218.7,221.1,216.9]

# (a) cetak pembacaan pertama dan terakhir memakai indeks
print("pembacaan pertama:",tegangan_harian[0])
print("pembacaan terakhir:",tegangan_harian[-1])

# (b) tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)
print(tegangan_harian)

# (c) cetak jumlah total jumlah data dengan len()
print("jumlah data:",len(tegangan_harian))
```
