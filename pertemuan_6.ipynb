```python
#6A

temps = [28, 31, 30, 33, 29]  # suhu (°C) selama 5 hari

print("temps[0] =", temps[0])                    # elemen pertama
print("temps[4] =", temps[4])                    # elemen terakhir (len - 1)
print("temps[-1] =", temps[-1])                  # indeks negatif: dihitung dari akhir

# --- prediksi dulu, lalu jalankan ---
print(temps[1])              # prediksi: 31
print(temps[-2])             # prediksi: 33
print(temps[len(temps) - 1]) # prediksi: 29
print(temps[2] + temps[3])  # prediksi: 63
```

```python
#6B

nums = [10, 20, 30]

nums.append(40)
nums.insert(1, 15)
nums.remove(30)

print("Final list:", nums)
print("Length:", len(nums))
```

```python
#6C

temps = [28, 31, 30, 33, 29]
print("valid indices: 0 to", len(temps) - 1)

try:
    print(temps[5])          # there is no box 5!
except IndexError as error:
    print("IndexError:", error)

# Safe ways to reach the last element, however long the list is
print("last:", temps[len(temps) - 1], "or simply", temps[-1])

# An empty list has no valid index at all
empty = []
print("empty list has", len(empty), "elements")
```
