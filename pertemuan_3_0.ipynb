```python
# Diketahui
jumlah_buku = 7
harga_per_buku = 18500
uang_budi = 150000
diskon = 0.10 # 10%

# 1. Total belanja, besar diskon, dan jumlah yang harus dibayar
total_belanja = jumlah_buku * harga_per_buku
besar_diskon = int(total_belanja * diskon)
bayar = total_belanja - besar_diskon

# 2. Uang kembalian
kembalian = uang_budi - bayar

# 3. Patungan 4 orang secara adil (bilangan bulat) dan sisanya
per_orang = bayar // 4
sisa = bayar % 4

# 4. Apakah uang Budi cukup?
uang_cukup = uang_budi >= bayar

# Menampilkan Output
print(f"Total       : {total_belanja}")
print(f"Diskon      : {besar_diskon}")
print(f"Bayar       : {bayar}")
print(f"Kembalian   : {kembalian}")
print(f"Per orang   : {per_orang}")
print(f"Sisa        : {sisa}")
print(f"Uang cukup? : {uang_cukup}")
```

```python
# Diketahui
jumlah_panel = 12
daya_per_panel = 450  # Wp
penyinaran = 5        # jam/hari
efisiensi = 0.8       # 80%
kebutuhan_harian = 25 # kWh/hari
tarif_listrik = 1444.70 # Rp/kWh
hari_per_bulan = 30

# 1. Energi yang dihasilkan per hari (kWh)
# Energi (Wh) = jumlah panel * daya * penyinaran * efisiensi
energi_harian_wh = jumlah_panel * daya_per_panel * penyinaran * efisiensi
energi_harian_kwh = energi_harian_wh / 1000  # Konversi Wh ke kWh

# 2. Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan
energi_bulanan_kwh = energi_harian_kwh * hari_per_bulan
nilai_listrik = energi_bulanan_kwh * tarif_listrik

# 3. Kekurangan energi per hari terhadap kebutuhan
kekurangan_kwh = kebutuhan_harian - energi_harian_kwh

# 4. Apakah PLTS sudah mencukupi kebutuhan rumah?
mencukupi = energi_harian_kwh >= kebutuhan_harian

# Menampilkan Output sesuai contoh
print(f"Energi harian (kWh) : {round(energi_harian_kwh, 1)}")
print(f"Energi bulanan (kWh): {round(energi_bulanan_kwh, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_listrik, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan_kwh, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
# Diketahui
daya_generator = 200    # kW
waktu_operasi = 8       # jam/hari
konsumsi_bbm = 0.25     # liter/kWh
harga_solar = 13500     # Rp/liter
kapasitas_tangki = 1000 # liter
arus = 20               # A
hambatan = 0.5          # Ohm

# 1. Energi yang dibangkitkan per hari (kWh)
energi = daya_generator * waktu_operasi

# 2. Kebutuhan solar per hari dan biayanya
solar_per_hari = energi * konsumsi_bbm
biaya_solar = solar_per_hari * harga_solar

# 3. Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
# Menggunakan floor division (//) dan modulo (%)
tangki_cukup = kapasitas_tangki // solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari

# 4. Rugi daya pada kabel (P = I^2 * R)
# Menggunakan operator eksponen (**)
rugi_kabel = (arus ** 2) * hambatan

# 5. Apakah rugi kabel tergolong tinggi?
# Menggunakan operator komparasi dan logika (>, >=, and)
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

# Menampilkan Output sesuai contoh
print(f"Energi (kWh)     : {energi}")
print(f"Solar (liter)    : {solar_per_hari}")
print(f"Biaya (Rp)       : {biaya_solar}")
print(f"Tangki cukup     : {tangki_cukup} hari")
print(f"Sisa solar (L)   : {sisa_solar}")
print(f"Rugi kabel (W)   : {rugi_kabel}")
print(f"Rugi tinggi?     : {rugi_tinggi}")
```

```python
# Deklarasi variabel
umur = 20           # Integer
berat_badan = 65.5  # Float
nama = "Hendi"      # String
angka = 255         # Integer (akan dikonversi ke heksa & oktal)

# 1. Operator %d (Integer)
print("Umur %s adalah %d tahun." % (nama, umur))

# 2. Operator %f (Float)
print("Berat badannya adalah %f kg." % berat_badan)
# (Opsional) %.2f digunakan untuk membatasi 2 angka di belakang koma
print("Berat badannya adalah %.2f kg." % berat_badan)

# 3. Operator %s (String)
print("Menu yang dipesan %s adalah Mi ayam bakso." % nama)

# 4. Operator %x (Hexadesimal)
print("Angka 255 dalam bentuk hexadesimal adalah %x." % angka)

# 5. Operator %o (Octal)
print("Angka 255 dalam bentuk octal adalah %o." % angka)
```
